  # 데이터수집
  1. 가격계열 (Price Features)
  코스피 지수 자체의 가격 변화를 나타내는 기본적/기술적 지표입니다.
   - 수집 데이터: 일별 KOSPI 지수 (Open, High, Low, Close)
   - 출처 및 라이브러리: yfinance (^KS11) 또는 FinanceDataReader (KS11)
   - 파생 변수 (Features): 
     - 수익률 (Return), 로그수익률 (Log Return)
     - 변동성 (Volatility, 20일 표준편차 등)
     - 보조지표: ATR (Average True Range), RSI (Relative Strength Index), MACD

  2. 거래량계열 (Volume Features)
  시장의 에너지와 모멘텀을 나타내는 거래량 기반 지표입니다.
   - 수집 데이터: 일별 KOSPI 거래량 (Volume), 시장 거래대금
   - 출처 및 라이브러리: yfinance, FinanceDataReader, pykrx
   - 파생 변수 (Features):
     - 거래량 변화율 (Volume Rate of Change)
     - VWAP (Volume Weighted Average Price)
     - Turnover (거래대금 회전율)

  3. 시장계열 (Market Features)
  코스피에 영향을 미치는 외부 글로벌/매크로 시장 지표입니다.
   - 수집 데이터 및 출처:
     - 글로벌 증시: S&P 500 수익률 (^GSPC via yfinance)
     - 금리: 미국 국채 10년물 금리 (^TNX via yfinance) 및 한국 국고채 금리
     - 환율/달러: 원/달러 환율 (USDKRW=X), 달러 인덱스 (DX-Y.NYB via yfinance)
     - 공포지수: VIX 지수 (^VIX via yfinance)

  4. 거시경제 (Macroeconomic Features)
  중장기적인 시장 환경을 나타내는 지표입니다. (주의: 주로 월간 단위 데이터이므로 일간 데이터와 병합 시 Forward Fill 처리 필요)
   - 수집 데이터 및 출처:
     - CPI (소비자물가지수): FRED API (미국 CPI) 또는 한국은행 ECOS API
     - 실업률 (Unemployment Rate): FRED API 또는 한국은행 ECOS API
     - PMI (구매관리자지수): ISM 제조업/비제조업 PMI (FRED 또는 Investing.com 등 크롤링)

  5. 비정형 데이터 (Unstructured Features)
  시장 심리를 나타내는 텍스트 및 검색 기반 데이터입니다.
   - 수집 데이터 및 출처:
     - 뉴스 Sentiment (감성 점수): 네이버 금융 주요 뉴스 크롤링 후 KoBERT 감성 분석 (긍정/부정 점수화)
     - 검색량: Naver DataLab API (예: "코스피", "폭락", "경제위기" 등 핵심 키워드 트렌드)
     - 공시 데이터: OpenDartReader를 활용한 시총 상위 주요 상장사의 호재/악재 공시 발생 건수

  6. 🆕 수급계열 (Supply & Demand Features)
  시장을 움직이는 핵심 주체들의 자금 흐름과 방향성입니다.
   - 수집 데이터: KOSPI 시장의 일자별, 투자자별 순매수 대금 (Net Purchases Value)
   - 출처 및 라이브러리: pykrx (get_market_trading_value_by_investor 또는 ETF 수급 함수 활용)
   - 타겟 추적 주체 (4개):
     1. 외국인 (Foreigner): 시장의 방향성을 주도하는 핵심 메이저 주체
     2. 금융투자 (Financial Investment): 증권사, 자산운용사 (고빈도/단기 방향성 및 차익거래 타겟)
     3. 투신 (Investment Trust): 일반 펀드 자금의 유출입 동향
     4. 연기금등 (Pension Fund): 국민연금 등 (장기적/방어적 시장 수급의 주체)
